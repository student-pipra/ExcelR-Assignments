# CHI-SQUARE TEST

### 1. State the Hypotheses:

###### H₀ (Null Hypothesis):
###### There is no association between the type of device and customer satisfaction level.

###### H₁ (Alternative Hypothesis):
###### There is an association between the type of device and customer satisfaction level.

###### H₀ → independent
###### H₁ → dependent

### 2. Compute the Chi-Square Statistic:

```python
import numpy as np
from scipy.stats import chi2_contingency

# Observed data (given table)
data = np.array([
    [50, 70],
    [80, 100],
    [60, 90],
    [30, 50],
    [20, 50]
])

# Chi-square test
chi2, p, dof, expected = chi2_contingency(data)

chi2
```

###### The Chi-Square test for independence is used to determine the association between two categorical variables.
###### The observed frequency table is created using the given data.
###### Expected frequencies are calculated based on row and column totals.
###### The Chi-Square statistic is then computed using the formula:
###### 
######  χ² = Σ [(O − E)² / E]
###### 
###### The calculated Chi-Square value is obtained using Python.

### 3. Determine the Critical Value:

```python
from scipy.stats import chi2

alpha = 0.05
df = 4

critical_value = chi2.ppf(1 - alpha, df)

critical_value
```

###### The critical value is determined using the Chi-Square distribution table.
###### It depends on the significance level (α = 0.05) and degrees of freedom (df = 4).
###### The critical value represents the threshold beyond which the null hypothesis is rejected.

### 4. Make a Decision:

###### The Chi-Square statistic is compared with the critical value at a significance level of 0.05.
###### If the Chi-Square value is greater than the critical value, the null hypothesis is rejected.
###### In this case, the calculated Chi-Square value is greater than the critical value.
###### Therefore, we reject the null hypothesis.
###### This indicates that there is a significant association between device type and customer satisfaction.

## Conclusion:

###### A Chi-Square test for independence was conducted to examine the relationship between device type and customer satisfaction.
###### The results showed that the calculated Chi-Square statistic was greater than the critical value at a 0.05 significance level.
###### Therefore, the null hypothesis was rejected.
###### This indicates that there is a significant association between the type of device purchased and customer satisfaction levels.

# HYPOTHESIS TESTING

### 1. State the Hypotheses statement:

###### H₀ (Null Hypothesis):
###### The weekly operating cost follows the theoretical model and is not higher than expected.
###### 
###### H₁ (Alternative Hypothesis):
###### The weekly operating cost is higher than the theoretical model suggests.

###### H₀ → no increase
###### H₁ → increase (right-tailed test)

### 2. Calculate the Test Statistic:

###### The theoretical weekly operating cost is given by:
###### W = 1000 + 5X
###### 
###### For X = 600 units:
###### μ = 1000 + 5 × 600 = 4000
###### 
###### Since population standard deviation is known, Z-test is used.

```python
import numpy as np

x_bar = 3050
mu = 1000 + 5 * 600
sigma = 5 * 25
n = 25

z_stat = (x_bar - mu) / (sigma / np.sqrt(n))

z_stat
```

### 3. Determine the Critical Value

###### Significance level (α) = 0.05
###### Right-tailed test
###### Critical value is taken from Z-distribution table.

```python
from scipy.stats import norm

z_critical = norm.ppf(1 - 0.05)

z_critical
```

### 4. Make a Decision

###### If Z-statistic > Z-critical → Reject Null Hypothesis
###### If Z-statistic ≤ Z-critical → Fail to Reject Null Hypothesis

```python
if z_stat > z_critical:
    decision = "Reject Null Hypothesis"
else:
    decision = "Fail to Reject Null Hypothesis"

decision
```

## Conclusion

###### The calculated Z-statistic is compared with the critical value at a 5% significance level.
###### Since the Z-statistic is greater than the critical value, we reject the null hypothesis.
###### This provides strong evidence that the weekly operating costs are higher than the theoretical model suggests.
