#### Histograms (Data Distribution)

```python
import matplotlib.pyplot as plt

df.hist(figsize=(15,10), bins=20)
plt.show()
```

###### Histograms show distribution of numerical columns.
###### They help to understand data shape (normal or skewed).
###### Useful for identifying patterns in data.

#### Boxplot (Outliers)

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12,6))
sns.boxplot(data=df)
plt.xticks(rotation=90)
plt.show()
```

###### Boxplot is used to detect outliers in numerical columns.
###### It shows median, quartiles, and extreme values.
###### Helps identify abnormal data points in dataset.

#### Correlation Heatmap

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10,8))
sns.heatmap(df.corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

###### Shows correlation between numerical variables.
###### Helps identify positive and negative relationships.
###### Useful for feature understanding in dataset.

### Scatter Plot (Relationships)

```python
import matplotlib.pyplot as plt

plt.scatter(df["LB"], df["ASTV"])
plt.xlabel("LB")
plt.ylabel("ASTV")
plt.title("LB vs ASTV Relationship")
plt.show()
```

###### Shows relationship between two numerical variables.
###### Helps identify patterns or correlation visually.
###### Useful for understanding feature interactions.

### Pie Chart (NSP Distribution)

```python
import matplotlib.pyplot as plt

df["NSP"].value_counts().plot.pie(autopct="%1.1f%%")
plt.title("NSP Distribution")
plt.ylabel("")
plt.show()
```

### Pairplot

```python
import seaborn as sns

sns.pairplot(df)
plt.show()
```

###### Shows pairwise relationships between variables.
###### Helps visualize patterns, clusters, and correlations.
###### Useful for deep data understanding.

### Violin Plot

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))
sns.violinplot(data=df)
plt.xticks(rotation=90)
plt.show()
```

###### Violin plot shows distribution and density of data.
###### It combines boxplot + distribution shape.
###### Helps understand data spread and skewness clearly.
