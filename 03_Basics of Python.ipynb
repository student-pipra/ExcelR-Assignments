## Exercise 1: Prime Numbers

```python
# Take input from user
num = int(input("Enter a number: "))

# Check if number is greater than 1
if num <= 1:
    print("Not Prime")
else:
    is_prime = True

    # Loop to check divisibility
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            is_prime = False
            break

    # Final result
    if is_prime:
        print("Prime Number")
    else:
        print("Not Prime")
```

## Exercise 2: Product of Random Numbers

```python
import random

# Generate two random numbers
num1 = random.randint(1, 10)
num2 = random.randint(1, 10)

print("Multiply:", num1, "and", num2)

# Take user input
user_answer = int(input("Enter your answer: "))

# Check if answer is correct
if user_answer == num1 * num2:
    print("Correct!")
else:
    print("Wrong! Correct answer is:", num1 * num2)
```

## Exercise 4: Word counter

```python
# Take input text from user
text = input("Enter text: ")

# Split text into words
words = text.split()

# Create dictionary to store word count
word_count = {}

# Count frequency of each word
for word in words:
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1

# Print result
for word, count in word_count.items():
    print(f"'{word}': {count}")
```

## Exercise 5: Check for Palindrome

```python
# Function to check palindrome
def is_palindrome(text):
    # Convert to lowercase and remove spaces
    text = text.lower().replace(" ", "")
    return text == text[::-1]

# Take input from user
user_input = input("Enter text: ")

# Check and print result
if is_palindrome(user_input):
    print("True")
else:
    print("False")
```

###### The input is converted to lowercase using the lower() function, so that case differences (like "Racecar" vs "racecar") do not affect the result. This ensures that the palindrome check is accurate and returns True.
